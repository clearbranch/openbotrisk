```python
from pathlib import Path
from urllib.request import urlretrieve
from ipaddress import ip_network

sources = {
    "vpn": "https://raw.githubusercontent.com/X4BNet/lists_vpn/main/output/vpn/ipv4.txt",
    "datacenter": "https://raw.githubusercontent.com/X4BNet/lists_vpn/main/output/datacenter/ipv4.txt",
}

for name, url in sources.items():
    subnet_file = Path(f"/tmp/{name}_subnets.txt")
    ip_file = Path(f"/tmp/{name}_ips.txt")

    urlretrieve(url, subnet_file)

    count = 0
    with subnet_file.open() as subnets, ip_file.open("w") as ips:
        for line in subnets:
            subnet = line.strip()
            if not subnet:
                continue

            network = ip_network(subnet, strict=False)
            for address in network:
                ips.write(f"{address}\n")
                count += 1

    print(f"{name}: wrote {count:,} IP addresses to {ip_file}")
```

```python
import urllib.request, hashlib, datetime
from pathlib import Path
from ipaddress import ip_network, ip_address
import numpy as np

RAW = Path("data/raw/x4bnet") / datetime.date.today().isoformat()
RAW.mkdir(parents=True, exist_ok=True)
URL = "https://raw.githubusercontent.com/X4BNet/lists_vpn/main/output/{}/ipv4.txt"

def fetch(name):
    data = urllib.request.urlopen(URL.format(name), timeout=60).read()
    (RAW / f"{name}_ipv4.txt").write_bytes(data)
    return hashlib.sha256(data).hexdigest()

def load_intervals(path):
    nets = sorted(ip_network(l.strip(), strict=False)
                  for l in path.read_text().splitlines() if l.strip())
    s = np.array([int(n.network_address) for n in nets], dtype=np.uint64)
    e = np.array([int(n.broadcast_address) for n in nets], dtype=np.uint64)
    assert (s[1:] > e[:-1]).all(), "overlapping ranges: collapse first"
    return s, e

def in_list(ips_u32, s, e):
    i = np.searchsorted(s, ips_u32, side="right") - 1
    out = np.zeros(len(ips_u32), bool)
    ok = i >= 0
    out[ok] = ips_u32[ok] <= e[i[ok]]
    return out
```

```python
import numpy as np, pandas as pd
from ipaddress import ip_address

def ip_to_u32(x):
    try:
        a = ip_address(x)
    except ValueError:
        return -1
    if a.version == 6:
        a = a.ipv4_mapped      # None for real IPv6
        if a is None:
            return -1
    return int(a)

def flag_ips(ip_series, s, e):
    uniq = pd.Series(ip_series.dropna().unique())   # look up each IP once
    codes = uniq.map(ip_to_u32).to_numpy(dtype=np.int64)
    flags = np.zeros(len(uniq), bool)
    ok = codes >= 0
    flags[ok] = in_list(codes[ok].astype(np.uint64), s, e)
    return ip_series.map(dict(zip(uniq, flags))).fillna(False).astype(bool)
```

```python
import hashlib, datetime, urllib.request
from pathlib import Path
import pandas as pd

RAW = Path("data/raw/lists") / datetime.date.today().isoformat()
RAW.mkdir(parents=True, exist_ok=True)

SOURCES = {
    "x4_vpn":        "https://raw.githubusercontent.com/X4BNet/lists_vpn/main/output/vpn/ipv4.txt",
    "x4_datacenter": "https://raw.githubusercontent.com/X4BNet/lists_vpn/main/output/datacenter/ipv4.txt",
    "cloud_merged":  "https://raw.githubusercontent.com/rezmoss/cloud-provider-ip-addresses/main/all_providers/all_providers_merged_v4.txt",
}

intervals, hashes = {}, {}
for name, url in SOURCES.items():
    data = urllib.request.urlopen(url, timeout=120).read()
    (RAW / f"{name}.txt").write_bytes(data)
    hashes[name] = hashlib.sha256(data).hexdigest()
    intervals[name] = load_intervals(RAW / f"{name}.txt")

# What's in each list
summary = pd.DataFrame({
    name: {"networks": len(s), "addresses": int((e - s + 1).sum())}
    for name, (s, e) in intervals.items()
}).T
summary["pct_of_ipv4"] = (summary["addresses"] / 2**32 * 100).round(2)
summary

# Check any IPs you like
def check(ips):
    ips = pd.Series(list(ips), name="ip")
    out = pd.DataFrame({"ip": ips})
    for name, (s, e) in intervals.items():
        out[name] = flag_ips(ips, s, e)
    out["checkable"] = ips.map(ip_to_u32) >= 0
    return out

check(["52.95.110.1", "8.8.8.8", "81.2.69.142", "2001:db8::1"])
```
